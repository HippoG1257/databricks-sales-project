```python
data=[
    (1,"laptop",2,50000),
    (2,"mobile",3,30000),
    (3,"Tablet",1,20000),
    (4,"laptop",2,50000)
]
columns = ["id","product","quantity","price"]
df = spark.createDataFrame(data,columns)
 
record_count = df.count()
print(f"Total records in the DataFrame: {record_count}")
if record_count == 10:
    raise Exception("No records found")
null_count=df.filter(df.quantity.isNull()).count()
print("Null quantity records",null_count)
 
```
